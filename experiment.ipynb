```python
import json
import datetime
```

```python
class abc:
    def __init__(self, i, cat, amt, desc, date):
        self.i = i
        self.cat = cat
        self.amt = amt
        self.desc = desc
        self.date = date
    def getItems(self):
        # return f'{self.i}, {self.cat}, {self.amt}, {self.desc}, {self.date}'
        return [self.i, self.cat, self.amt, self.desc, self.date]
exp = abc(1, "Food", 2000, "Pizza", datetime.date.today().isoformat())
```

```python
exp_dict = {
    'Id':exp.i,
    'Category':exp.cat,
    'Amount':exp.amt,
    'Description':exp.desc,
    'Date':exp.date
}
json.dumps(exp_dict, separators=(',',':'))
```
